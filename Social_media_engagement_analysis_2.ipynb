```python
from google.colab import files
import pandas as pd
import numpy as np

# Upload dataset
uploaded = files.upload()

# Read CSV file
file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

# Clean column names
df.columns = df.columns.str.strip().str.replace('\ufeff', '')

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)
print("Available Columns:", df.columns.tolist())

display(df.head())
```

```python
import numpy as np

# Creating NumPy Arrays
likes = np.array(df['likes'])
comments = np.array(df['comments'])
shares = np.array(df['shares'])

print("Likes Array:")
print(likes)

print("\nComments Array:")
print(comments)

print("\nShares Array:")
print(shares)

# Array Indexing
print("\nArray Indexing:")
print("First Like:", likes[0])
print("Second Like:", likes[1])
print("Third Like:", likes[2])
print("Last Like:", likes[-1])
```

```python
# Array Slicing
print("First 10 Likes:")
print(likes[:10])

print("\nLikes from 10 to 20:")
print(likes[10:20])

print("\nEvery 5th Like:")
print(likes[::5])

# NumPy Operations
print("\nAddition:")
print(likes[:10] + comments[:10])

print("\nSubtraction:")
print(likes[:10] - comments[:10])

print("\nMultiplication:")
print(likes[:10] * comments[:10])

print("\nDivision:")
print(likes[:10] / (comments[:10] + 1))

# Statistical Operations
print("\nMaximum Likes:", np.max(likes))
print("Minimum Likes:", np.min(likes))
print("Average Likes:", np.mean(likes))
print("Standard Deviation:", np.std(likes))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 3.1 Data Distribution
print("Likes Distribution:")
print(df['likes'].describe())

plt.hist(df['likes'], bins=10)
plt.title("Distribution of Likes")
plt.xlabel("Likes")
plt.ylabel("Frequency")
plt.show()

# 3.2 Descriptive Statistics
print("\nDescriptive Statistics:")
display(df.describe())

print("Average Likes:", np.mean(df['likes']))
print("Maximum Likes:", np.max(df['likes']))
print("Minimum Likes:", np.min(df['likes']))

# 3.3 Correlation and Covariance
numeric_df = df.select_dtypes(include='number')

print("\nCorrelation:")
display(numeric_df.corr())

print("\nCovariance:")
display(numeric_df.cov())

sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
# 3.4 Outlier Detection using IQR
Q1 = df['likes'].quantile(0.25)
Q3 = df['likes'].quantile(0.75)

IQR = Q3 - Q1

outliers = df[
    (df['likes'] < Q1 - 1.5 * IQR) |
    (df['likes'] > Q3 + 1.5 * IQR)
]

print("Number of Outliers:", len(outliers))
display(outliers.head())

sns.boxplot(x=df['likes'])
plt.title("Outlier Detection in Likes")
plt.show()

# 3.5 Feature Relationships
sns.scatterplot(data=df, x='views', y='likes', hue='platform')

plt.title("Views vs Likes")
plt.xlabel("Views")
plt.ylabel("Likes")
plt.show()

sns.scatterplot(data=df, x='comments', y='shares')

plt.title("Comments vs Shares")
plt.xlabel("Comments")
plt.ylabel("Shares")
plt.show()

sns.scatterplot(data=df, x='reach', y='engagement_rate_percent')

plt.title("Reach vs Engagement Rate")
plt.xlabel("Reach")
plt.ylabel("Engagement Rate (%)")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 4.1 Platform Comparison
platform_analysis = df.groupby('platform').agg({
    'likes': 'mean',
    'comments': 'mean',
    'shares': 'mean',
    'engagement_rate_percent': 'mean'
}).round(2)

print("Platform Performance:")
display(platform_analysis)

platform_analysis['engagement_rate_percent'].plot(
    kind='bar', figsize=(7, 4)
)

plt.title("Average Engagement Rate by Platform")
plt.xlabel("Platform")
plt.ylabel("Engagement Rate (%)")
plt.xticks(rotation=0)
plt.show()

# 4.2 Content Type Analysis
content_analysis = df.groupby('content_type').agg({
    'likes': 'mean',
    'comments': 'mean',
    'shares': 'mean',
    'views': 'mean'
}).round(2)

print("\nContent Type Performance:")
display(content_analysis)

sns.barplot(
    data=df,
    x='content_type',
    y='engagement_rate_percent'
)

plt.title("Engagement Rate by Content Type")
plt.xlabel("Content Type")
plt.ylabel("Engagement Rate (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
# 4.3 Monthly Engagement Trend
df['post_date'] = pd.to_datetime(df['post_date'], errors='coerce')

df['month'] = df['post_date'].dt.to_period('M').astype(str)

monthly_analysis = df.groupby('month')['engagement_rate_percent'].mean()

print("Monthly Engagement:")
display(monthly_analysis)

monthly_analysis.plot(kind='line', marker='o', figsize=(9, 4))

plt.title("Monthly Engagement Trend")
plt.xlabel("Month")
plt.ylabel("Average Engagement Rate (%)")
plt.xticks(rotation=45)
plt.show()

# 4.4 Topic Analysis
topic_analysis = df.groupby('topic').agg({
    'likes': 'mean',
    'comments': 'mean',
    'shares': 'mean',
    'engagement_rate_percent': 'mean'
}).round(2)

print("\nTopic-wise Performance:")
display(topic_analysis)

topic_analysis['engagement_rate_percent'].sort_values().plot(
    kind='barh', figsize=(8, 5)
)

plt.title("Engagement Rate by Topic")
plt.xlabel("Average Engagement Rate (%)")
plt.ylabel("Topic")
plt.show()
```

```python
# 5.1 Top Performing Content

top_posts = df.sort_values(
    by='engagement_rate_percent',
    ascending=False
)

print("Top 10 Performing Posts:")
display(top_posts[
    ['post_id', 'platform', 'content_type',
     'likes', 'comments', 'shares',
     'engagement_rate_percent']
].head(10))

# 5.2 Platform Performance

platform_result = df.groupby('platform')[
    'engagement_rate_percent'
].mean().sort_values(ascending=False)

print("\nPlatform Performance:")
display(platform_result)

print("Best Performing Platform:", platform_result.idxmax())
```
